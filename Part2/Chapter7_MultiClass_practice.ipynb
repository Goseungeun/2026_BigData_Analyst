## 신용 등급 예측

은행 정보로 신용 등급을 예측하시오.
- 제공된 데이터 목록 : score_train.csv , score_test.csv
- 예측할 컬럼 : Credit_Score(Good,Standard,Poor)

학습용 데이터를 이용해 신용 등급을 예측하는 모델을 만든 후 이를 평가용 데이터에 적용해 얻은 예측값을 다음과 같은 형식의 csv파일로 생성하시오.

제출 파일은 다음 1개의 컬럼을 포함해야 한다.
- pred : 예측값
- 제출 파일명 : 'result.csv'

제출한 모델의 성능은 f1-macro 평가지표에 따라 채점한다.

```python
import pandas as pd

train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part2/ch7/score_train.csv")
test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part2/ch7/score_test.csv")

#EDA
# print(train.shape , test.shape)
# print(train.info())
# print(train.isnull().sum().sum())
# print(test.isnull().sum().sum())

#Data preprocessing
target = train.pop('Credit_Score')

train = pd.get_dummies(train)
test = pd.get_dummies(test)

#split train val set
from sklearn.model_selection import train_test_split

X_train,X_val,y_train,y_val = train_test_split(train,target,test_size = 0.2,random_state = 42)

# train and validation
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score

rf = RandomForestClassifier()
rf.fit(X_train,y_train)
pred = rf.predict(X_val)

#answer file creation
pred = rf.predict(test)
ans = pd.DataFrame({'pred':pred})
ans.to_csv('result.csv',index=False)

print(pd.read_csv('result.csv').head())
```

## 약물 종류 예측

- 주어진 데이터에서 약물의 종류를 예측하시오.
  - 제공된 데이터 목록 : drug_train.csv, drug_test.csv
  - 예측할 컬럼 : Drug
- 학습용 데이터를 이용해 약물의 종류를 예측하는 모델을 만든 후 이를 평가용 데이터에 적용해 얻은 예측값을 다음과 같은 형식의 csv 파일로 생성하시오.

제출파일은 다음 1개 컬럼을 포함해야 한다.
- pred : 예측값
- 제출 파일명 : 'result.csv'

제출한 모델의 성능은 f1-macro 평가지표에 따라 채점한다.


```python
import pandas as pd

train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part2/ch7/drug_train.csv")
test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part2/ch7/drug_test.csv")

#EDA
# print(train.shape,test.shape)
# print(train.isnull().sum().sum())
# print(test.isnull().sum().sum())
# print(train.nunique())
# print(test.nunique())

#preprocessing
target = train.pop('Drug')

train = pd.get_dummies(train)
test = pd.get_dummies(test)

#split
from sklearn.model_selection import train_test_split
x_train,x_val,y_train,y_val = train_test_split(train,target,test_size = 0.2,random_state = 42)

#trian, validation
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(x_train,y_train)
pred = rf.predict(x_val)

from sklearn.metrics import f1_score
print(f1_score(y_val,pred,average='macro'))

# answer file creation
pred = rf.predict(test)
ans = pd.DataFrame({'pred':pred})
ans.to_csv('result.csv',index=False)

print(pd.read_csv('result.csv').head())
```

## 유리종류 예측

- 유리 식별 데이터에서 유리의 종류를 예측하시오
  - 제공된 데이터 목록 : glass_train.csv, glass_test.csv
  - 예측할 컬럼 : Type
- 학습용 데이터를 이용해 약물 종류를 예측하는 모델을 만든 후 이를 평가용 데이터에 적용해 얻은 예측값을 다음과 같은 형식의 CSV 파일로 생성하시오

제출파일은 다음 1개의 컬럼을 포함해야 한다.
- pred : 예측값
- 제출 파일명: 'result.csv'

제출한 모델의 성능은 f1-weighted 평가지표에 따라 채점한다.


```python
import pandas as pd

train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part2/ch7/glass_train.csv")
test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part2/ch7/glass_test.csv")

# #EDA
# print(train.shape, test.shape)
# print(train.isnull().sum().sum())
# print(test.isnull().sum().sum())
# print(train.info())

#preprocess and split

from sklearn.model_selection import train_test_split
target = train.pop('Type')

x_train,x_val,y_train,y_val = train_test_split(train,target,test_size=0.2,random_state=0)

#train and validation

from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(x_train,y_train)
pred = rf.predict(x_val)

from sklearn.metrics import f1_score
print(f1_score(y_val,pred,average='weighted'))

#result file creation
pred = rf.predict(test)
ans = pd.DataFrame({'pred':pred})
ans.to_csv('result.csv',index=False)

print(pd.read_csv('result.csv').head())
```
