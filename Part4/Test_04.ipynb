```python
# Q2
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch4/data4-2.csv')

df['ratio'] = ((df['loves'] + df['wows'])/df['reactions']) * 100

cond = (df['ratio'] > 40) & (df['ratio'] < 50)
cond2 = df['type'] == 'video'
print(len(df[cond & cond2]))
```

```python
# Q3
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch4/data4-3.csv')

df['date_added'] = pd.to_datetime(df['date_added'])
df['year'] = df['date_added'].dt.year
df['month'] = df['date_added'].dt.month

cond1 = (df['year'] == 2018) & (df['month'] == 1)
cond2 = df['country'] == 'United Kingdom'

print(len(df[cond1 & cond2]))
```

```python
import pandas as pd
train = pd.read_csv('https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch4/train.csv')
test = pd.read_csv('https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch4/test.csv')

print(train.shape)
print(test.shape)

print(train.info())
print(test.info())


cols = test.columns[test.dtypes == 'object'].tolist()
for col in cols:
  train_nu = train[col].nunique()
  test_nu = test[col].nunique()
  if train_nu != test_nu:
    print(col,'not same!!!!!!')
  else:
    print(col,'same num:',train_nu)

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score

target = train.pop('Segmentation')
train = pd.get_dummies(train)
test = pd.get_dummies(test)

x_train,x_val,y_train,y_val=train_test_split(train,target,test_size=0.2,random_state=42)
rf = RandomForestClassifier()
rf.fit(x_train,y_train)
val_pred = rf.predict(x_val)

# print(f1_score(y_val,val_pred,average='macro'))

pred = rf.predict(test)
result = pd.DataFrame({'pred':pred})
result.to_csv('result.csv',index=False)

print(pd.read_csv('result.csv').shape)
print(pd.read_csv('result.csv').head())
```
