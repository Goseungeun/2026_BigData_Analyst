```python
#Q2
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/lovedlim/bae-v3/main/part4/ch11/sensor_data.csv')

# 2-1
print(df.isnull().sum()) # S2_Temp

mid = df['S2_Temp'].median()
# 2-2
df['S2_Temp'] = df['S2_Temp'].fillna(mid)

# 2-3
print(round(df['S2_Temp'].mean(),3)) #25.055
```

```python
# Q3
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bae-v3/main/part4/ch11/order_data.csv")

df['order_amt'] = df['quantity'] * df['price']
df['cancel_TF'] = df['order_no'].str.startswith('C')

cond = df['cancel_TF'] == True
df.loc[cond,'order_amt'] = df.loc[cond,'order_amt']*(-1)

# 3-1
filtered_df = df[cond]
filtered_df = filtered_df.sort_values('order_amt')
print(filtered_df.iloc[0]) #C542426

# 3-2
grouped_df = df.groupby('cus_id').sum()
print(grouped_df['order_amt'].max())
#2877.21
```

```python
import pandas as pd
train = pd.read_csv('https://raw.githubusercontent.com/lovedlim/bae-v3/main/part4/ch11/game_train.csv')
test = pd.read_csv('https://raw.githubusercontent.com/lovedlim/bae-v3/main/part4/ch11/game_test.csv')
#target = User_Level

#EDA
print(train.isnull().sum().sum())
print(test.isnull().sum().sum())

print(train.shape)
print(test.shape)

print(train.info())

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
from sklearn.model_selection import train_test_split

target = train.pop('User_Level')
X_train,X_val,y_train,y_val = train_test_split(train,target,test_size=0.2,random_state=0)

rf = RandomForestClassifier()
rf.fit(X_train,y_train)
pred = rf.predict(X_val)

print(f1_score(y_val,pred,average = 'macro'))

ans = rf.predict(test)
result = pd.DataFrame({'pred':ans})
result.to_csv('result.csv',index=False)
print(pd.read_csv('result.csv').head())
```

```python
#Q1

import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/lovedlim/bae-v3/main/part4/ch11/elderly_health.csv')

train = df.iloc[:2000]
test = df.iloc[2000:]

import numpy as np
from statsmodels.formula.api import logit

model = logit("predicted ~ age+diabetic+activity+glus_fast+bmi+blood_pressure",data=train).fit()
print(model.summary())
# 1-1 정답 : 0.36
odds = np.exp(model.params['diabetic[T.yes]'])
print(round(odds,2))

# 1-2 정답 : 181
pred = model.predict(test)
m = pred.idxmax()
print(test.loc[m,'glus_fast'])

#1-3 정답 : 0.94
pred = (pred > 0.2).astype(int)

from sklearn.metrics import recall_score
s = recall_score(test['predicted'],pred)
print(round(s,2))
```

```python
# Q2
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bae-v3/main/part4/ch11/promotion_data.csv")

# 2-1
from scipy.stats import ttest_1samp
t_stat, p_value = ttest_1samp(df['sales'],35000)
print(round(p_value,3)) #0.168

# 2-2
print(df.corr()) # 0.575396

# 2-3
from statsmodels.formula.api import ols
model = ols('sales~visit_count+page_time+ad_clicks+promotion_budget',data = df).fit()
ans = model.pvalues.idxmin()
print(round(model.params[ans],3)) # 130.739
```
