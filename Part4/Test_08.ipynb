## Part2

```python
import pandas as pd
train = pd.read_csv('https://raw.githubusercontent.com/lovedlim/bae-v3/main/part4/ch8/churn_train.csv')
test = pd.read_csv('https://raw.githubusercontent.com/lovedlim/bae-v3/main/part4/ch8/churn_test.csv')

print(train.shape)
print(test.shape)

print(train.info())
print(train.isnull().sum().sum())
print(test.isnull().sum().sum())

# print(train.nunique())
# print(test.nunique())

#preprocessing
train = train.drop('customerID',axis=1)
test = test.drop('customerID',axis=1)

target = train.pop('TotalCharges')
train = pd.get_dummies(train)
test = pd.get_dummies(test)

print(train.shape)
print(test.shape)

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

x_train,x_val,y_train,y_val = train_test_split(train,target,test_size = 0.2,random_state=0)
rf = RandomForestRegressor()
rf.fit(x_train,y_train)
val_pred = rf.predict(x_val)

pred = rf.predict(test)
result = pd.DataFrame({'pred':pred})
result.to_csv('result.csv',index=False)

print(pd.read_csv('result.csv').head())
print(pd.read_csv('result.csv').shape)
```

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/lovedlim/bae-v3/main/part4/ch8/churn.csv')
print(df.columns) # 'AccountWeeks', 'ContractRenewal', 'DataPlan', 'DataUsage','CustServCalls', 'DayMins', 'DayCalls', 'MonthlyCharge', 'OverageFee','RoamMins'
#1-1
from statsmodels.formula.api import logit
model = logit('Churn ~ AccountWeeks+ContractRenewal+DataPlan+DataUsage+CustServCalls+DayMins+DayCalls+MonthlyCharge+OverageFee+RoamMins',data=df).fit()
print(model.summary()) #8개

#1-2 DataUsage DayMins
model2 = logit('Churn ~ DataUsage + DayMins',data = df).fit()
print(round(model2.params.sum(),3)) # -1.213

#1-3
import numpy as np
print(np.exp(model2.params['DataUsage']*5)) #0.428
```

```python
#Q2
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/lovedlim/bae-v3/main/part4/ch8/piq.csv')

from statsmodels.formula.api import ols
print(df.info())
model = ols('PIQ~Brain+Height+Weight',data=df).fit()
# 2-1
print(model.summary()) #2.343

#2-2
print(round(model.rsquared,2)) #0.37

#2-3
new = pd.DataFrame({'Brain':[90],'Height':[70],'Weight':[150]})
pred = model.predict(new)
print(round(pred[0])) #106
```
