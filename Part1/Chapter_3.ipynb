## Section 1 필터링, 최솟값, 중앙값

1. 'f5'컬럼이 0이 아닌 데이터(행)을 구하시오.
2. 앞에서 구한 데이터에 'views' 컬럼 결측치를 'views'컬럼의 최솟값으로 채워주세요.
3. 그리고 'views'컬럼의 중앙값을 계산해 정수로 구하시오.

```python
import pandas as pd
pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',None)
pd.set_option('display.float_format','{:.10f}'.format)


df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part1/ch3/type1_data1.csv")

# 0 제외
cond = df['f5'] != 0
df = df[cond]

# 결측치 최솟값으로 채우기
# print(dir(df))
views_min = df['views'].min()
df['views'] = df['views'].fillna(views_min)

# 중앙값 계산 -> 정수
views_mid = df['views'].median()
print(round(views_mid))
```

## Section 16
1. 'views'컬럼의 1사분위수, 3사분위수 그리고 IQR을 계산하시오.
2. 이상치 조건에 맞는 데이터를 찾으시오. (이상치는 1사분위수 - (IQR*1.5)보다 작은 값과 3사분위수 + (IQR*1.5)보다 큰 값)
3. 이상치 데이터의 'views'컬럼 합을 정수로 구하시오.

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part1/ch3/type1_data1.csv")

q3 = df['views'].quantile(.75)
q1 = df['views'].quantile(.25)

IQR = q3 - q1

line1 = q1 - IQR*1.5
line2 = q3 + IQR*1.5

cond1 = df['views'] < line1
cond2 = df['views'] > line2
df = df[cond1|cond2]

print(int(df['views'].sum()))
```

## Section 22

1. 주문 시간과 실제 도착 시간의 차이를 분 단위로 계산하시오.
2. 앱 종류별로 평균 도착시간(분)을 계산하시오.
3. 평균적으로 가장 빠른 앱 종류를 찾고, 해당 앱의 평균 도착 시간을 분으로 반올림하여 정수로 구하시오.

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part1/ch3/delivery_time.csv")

df['실제도착시간'] = pd.to_datetime(df['실제도착시간'])
df['주문시간'] = pd.to_datetime(df['주문시간'])

df['diff'] = (df['실제도착시간'] - df['주문시간']).dt.total_seconds()/60

df = df.groupby('앱종류')['diff'].mean().reset_index()

df = df.sort_values('diff')
print(round(df.iloc[0,1]))
```

## Section 23

1. 각 결제 종류별로 실제 도착 시간이 예상 도착 시간보다 늦은 주문의 비율을 계산하시오.
2. 비율 중 가장 큰 값을 반올림하여 소수 둘째 자리까지 구하시오.

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part1/ch3/delivery_time.csv")

df['실제도착시간'] = pd.to_datetime(df['실제도착시간'])
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'])

df['지연시간'] = (df['실제도착시간'] - df['예상도착시간']).dt.total_seconds()/60

df['지연여부'] = df['지연시간']>0

result = df.groupby('결제종류')['지연여부'].mean()

print(round(result.max(),2))
```

## Section 27

1. 점심시간(10시부터 13시 전까지) 주문된 배달 데이터를 찾으시오.
2. 점심시간 주문 건 중 과속 (평균 속도가 50km/h 이상)하는 주문 수를 정수로 구하시오.
  - 배달시간 = 실제도착시간 - 주문시간
  - 속도 (km/h)= 거리 (km)/시간(h)

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part1/ch3/delivery_time.csv")
'''
주문시간, 실제도착시간-> datetime
주문시간 => 시간 뽑기
시간 10~13 사이 찾기
배달시간 = 실제도착시간 - 주문시간
속도 = 거리/배달시간
'''
df['주문시간'] = pd.to_datetime(df['주문시간'])
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'])

df['주문시각'] = df['주문시간'].dt.hour

con1 = (df['주문시각'] >=10) & (df['주문시각'] <13)
df = df[con1]

df['배달시간'] = df['실제도착시간'] - df['주문시간']
df['배달시간'] = df['배달시간'].dt.total_seconds()/60/60
df['속도'] = df['거리'] / df['배달시간']

con2 = df['속도'] >= 50
df = df[con2]

print(int(len(df)))
```

## section 29

1. 배달료 계산 기준표에 따라 각 주문에 대한 배달료를 계산하시오.
2. 연도-월별로 배달료의 총합을 집계하시오.
3. 배달료가 가장 많이 발생한 월을 찾고, 그 월의 총 배달료를 정수로 구하시오.

[배달료 기준표]
- 5km 미만 : 2,000원
- 5km 이상 10km 미만 : 4,000원
- 10km 이상 15km 미만 : 6,000원
- 15km 이상 20km 미만 ; 8,000원

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part1/ch3/delivery_time.csv")
'''
배달료 계산 함수 작성 후 df 에 적용
주문시간 datetime 변경
주문시간 연월 구분 dt.to_period('M')
연도-월 별 배달료 총합 집계 groupby.sum()
배달료 상위 발생 월 -> 그 월의 총 배달료 정수로 구하기
'''
def cal_delievery_fee(distance):
  if distance < 5:
    return 2000
  elif distance < 10:
    return 4000
  elif distance <15:
    return 6000
  elif distance < 20:
    return 8000

df['fee'] = df['거리'].apply(cal_delievery_fee)

df['주문시간'] = pd.to_datetime(df['주문시간'])
df['order_ym'] = df['주문시간'].dt.to_period('M')

fee_df = df.groupby('order_ym')['fee'].sum()
print(int(fee_df.max()))
```

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/refs/heads/main/part1/ch3/store_sales.csv")

df['매출액'] = df['판매수량'] * df['단가']

day_map = {'월':'week','화':'week','수':'week','목':'week','금':'week','토':'weekend','일':'weekend'}
df['weekday'] = df['요일'].map(day_map)

store_sales = df.groupby(['매장코드','weekday'])['매출액'].sum().unstack()
store_sales['차이'] = abs(store_sales['week'] - store_sales['weekend'])
store_sales['차이'].max()
```

## Section 38

1. 각 region과 Channel 조합별로 제품(Product) A,B의 총 판매액을 계산하시오.
2. 제품 A의 매출 비율을 구하시오. A비율 = (제품 A판매액) / (제품 A 판매액 + 제품 B 판매액)
3. A비율 중 최대값을 찾아 반올림하여 소수 둘째 자리까지 구하시오.

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/refs/heads/main/part1/ch3/region_sales.csv")

pv = pd.pivot_table(df,index=['Region','Channel'],columns='Product', values='Sales',aggfunc='sum')

pv['total'] = pv['A'] + pv['B']
pv['A_ratio'] = pv['A'] / pv['total']

ans = pv['A_ratio'].max()
print(round(ans,2))
```
