## Section 1 필터링, 최솟값, 중앙값

1. 'f5'컬럼이 0이 아닌 데이터(행)을 구하시오.
2. 앞에서 구한 데이터에 'views' 컬럼 결측치를 'views'컬럼의 최솟값으로 채워주세요.
3. 그리고 'views'컬럼의 중앙값을 계산해 정수로 구하시오.

```python
import pandas as pd
pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',None)
pd.set_option('display.float_format','{:.10f}'.format)


df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part1/ch3/type1_data1.csv")

# 0 제외
cond = df['f5'] != 0
df = df[cond]

# 결측치 최솟값으로 채우기
# print(dir(df))
views_min = df['views'].min()
df['views'] = df['views'].fillna(views_min)

# 중앙값 계산 -> 정수
views_mid = df['views'].median()
print(round(views_mid))
```

## Section 3

1. 결측치가 있는 데이터(행)을 제거하시오.
2. 'views'컬럼을 'f1' 컬럼으로 나눈 값을 새로운 컬럼으로 추가하시오.
3. 새로운 컬럼 값 중 가장 큰 값을 가진 행의 age를 정수로 구하시오.

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part1/ch3/type1_data1.csv")

# 결측치 제거
df = df.dropna()

# 신규 컬럼 추가
df['new'] = df['views']/df['f1']

#new 컬럼 기준 내림차순
df = df.sort_values('new',ascending = False)
print(int(df.iloc[0,1]))
```

## Section 4
1. 'views'컬럼의 결측 데이터를 0으로 대체하시오.
2. 'views'컬럼에서 상위 10번째 값을 구하시오.
3. 'views'컬럼에서 상위 10개의 값을 상위 10번째 값으로 대체하시오.
4. 'views'컬럼의 전체 합을 정수로 구하시오.

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part1/ch3/type1_data1.csv")

#결측치 0 대체
df['views'] = df['views'].fillna(0)

# views 컬럼 기준 내림차순 정렬
df = df.sort_values('views',ascending=False)

# 상위 10번째 값 대입
tenth_views = df.iloc[9]['views']
df.iloc[:10,-1] = tenth_views

#views 컬럼 전체 합
print(int(df['views'].sum()))
```

## Section 7
1. 모든 나이 (age)에 1을 더하시오.
2. 20대의 'views'평균과 30대의 'views'평균의 차이의 절댓값을 구하시오.(반올림 후 소수 둘째 자리까지 계산)

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part1/ch3/type1_data1.csv")

# 1 추가
df['age'] = df['age'] + 1

#20대, 30대 조건
cond_20 = (df['age'] >= 20) & (df['age'] <30)
cond_30 = (df['age'] >= 30) & (df['age'] <40)

df_20 = df[cond_20]
df_30 = df[cond_30]

views_20_mean = df_20['views'].mean()
views_30_mean = df_30['views'].mean()

print(round(abs(views_20_mean - views_30_mean),2))
```

## Section 11
1. 데이터에서 결측치가 있는 데이터(행)을 모두 제거하시오.
2. 결측치가 제거된 데이터를 사용하여 앞에서부터 70% 데이터를 구하시오. (단, 데이터 70% 지점의 index가 소수점으로 계산될 경우 소수점 이하는 버림)
3. 앞에서 구한 70% 데이터 중 'views'컬럼의 3사분위수에서 1사분위수를 뺀 값을 소수점 이하를 버리고 정수 부분만 구하시오.

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part1/ch3/type1_data1.csv")

#결측치 제거
df = df.dropna()

#앞에서 70% 데이터
end = int(len(df) * 0.7)

df=df.iloc[:end]

#views컬럼의 3사분위수에서 1사분위를 뺀 값
q3 = df['views'].quantile(.75)
q1 = df['views'].quantile(.25)

print(int(q3 - q1))
```
